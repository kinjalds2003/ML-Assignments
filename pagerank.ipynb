```python
import numpy as np
temp_sum=[]
def linkinpark(adj_mat,pr, row, column):
    temp_sum.append(pr[row]/np.count_nonzero(adj_mat[row]))
    return sum(temp_sum)
```

PR(i)=(1-d)/N + d* summation over Pj belong to M(Pi)(PR(Pj)/L(Pj))

d: damping factor
N: number of pages
L(Pj): number of outgoing links from page Pj
M(Pi): set of all webpages linking to Pi

initilly all have same page rank of 1/N

```python
adj_mat=np.array([
    [0,1,1,0],
    [0,0,1,0],
    [1,0,0,0],
    [0,0,1,0]
])
pr=np.array([0.25,0.25,0.25,0.25])
eps=0.01
iteration=0
d=0.85
updated_pr=np.zeros(4)
nodes=['a','b','c','d']
out_degrees = np.count_nonzero(adj_mat, axis=1)
while True:
    for row in range(4):
        rank_sum=0
        for column in range(4):
           if adj_mat[column][row]==1:
                rank_sum+=pr[column]/out_degrees[column]

        updated_pr[row]=((1-d)/4)+(d*rank_sum)
    if np.all(np.abs(updated_pr - pr) < eps):
        pr = updated_pr
        break
    pr = updated_pr
    iteration += 1
    print(iteration)
print(pr)
```
