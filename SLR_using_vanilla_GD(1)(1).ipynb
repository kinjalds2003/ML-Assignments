```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/drive/MyDrive/ML @ SP for MSc DS 2024-2026/Synthetic.csv')
```

```python
df.head()
```

```python
df.drop(columns=["Unnamed: 0"], inplace=True)
```

```python
df.head()
```

```python
# Extract features and labels
x = df.iloc[:, 0].values
y = df.iloc[:, 1].values
```

```python
x.shape, y.shape
```

```python
x = (x - np.mean(x)) / np.std(x)
```

```python
x = x.reshape(1, -1)
y = y.reshape(-1, 1)
```

```python
x.shape, y.shape
```

```python
noOfFeatures, m = x.shape
```

```python
# Initialize model parameters
theta_0 = np.array(np.random.randn()).reshape(1,1)
theta_1 = np.array(np.random.randn()).reshape(1,1)
```

```python
theta_0.shape, theta_1.shape
```

```python
learning_rate = 0.0001
epochs = 10000
tolerance = 0.001
```

```python
epoch = 1
loss = 1
losses = []

while epoch <= epochs and loss > tolerance:

    # Forward pass
    y_hat = theta_0 * x + theta_1

    # Loss calculation
    loss = 1/2 * np.mean((y_hat.T - y)**2)

    # derivative calculation
    d_theta_0 = 1/m * np.dot(x, (y_hat.T - y))
    d_theta_1 = 1/m * np.sum(y_hat.T - y)

    # Update parameters
    theta_0 = theta_0 - learning_rate * d_theta_0
    theta_1 = theta_1 - learning_rate * d_theta_1

    losses.append(loss)

    epoch += 1
```

```python
print(theta_0.shape, theta_1.shape)
```

```python
plt.plot(losses)
plt.xlabel("Number of epochs")
plt.ylabel("Loss (MSE)")
```
